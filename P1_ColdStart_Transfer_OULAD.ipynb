# Cold-start learner-outcome prediction on OULAD

Run this notebook from the repository root. By default it reads the archived aggregate results. Set the raw data path and enable the run below to regenerate all results. The notebook calls the same pipeline as the command-line workflow.

```
from pathlib import Path
import subprocess
import sys

import pandas as pd

RUN_EXPERIMENTS = False
RAW = Path('data_oulad')  # extracted CSV directory or OULAD ZIP
if RUN_EXPERIMENTS:
    subprocess.run([sys.executable, 'run_pipeline.py', '--raw', str(RAW)], check=True)
results = Path('results' if RUN_EXPERIMENTS else 'reference-results')
```

## Benchmark and adaptation

Each target is tested on 2014J after removing its test learners from the source and target pool.

```
display(pd.read_csv(results / 'domain_shift.csv').round(3))
display(pd.read_csv(results / 'main_all.csv').groupby(['k', 'method'])[['succ_auc', 'wd_auc', 'wd_ap']].mean().round(3))
display(pd.read_csv(results / 'adapt_vs_zeroshot.csv').round(3))
```

## Uncertainty and follow-up analyses

Runs with different seeds for one module reuse the test cohort. Seed-level tests are exploratory. The day-28 DML estimate is an adjusted association for this analysis and does not establish an intervention effect.

```
for file in ['wilcoxon.csv', 'bootstrap_head_vs_lgbmP.csv', 'bayes_signed_rank.csv',
             'ablation.csv', 'stream.csv', 'anytime_with70.csv',
             'shap_top10.csv', 'fairness_summary.csv', 'causal_dml.csv']:
    print(file)
    display(pd.read_csv(results / file).round(3).head(10))
```
